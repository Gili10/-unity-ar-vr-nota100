METODOLOGIA

**4.1 Conjunto de Dados**
Foram utilizadas imagens do dataset C-NMC (Cancer vs. Non-Cancer), composto por amostras de células sanguíneas digitalizadas, classificadas em duas categorias: ALL (leucemia linfoblástica aguda) e HEM (células normais). As imagens foram redimensionadas para 224×224 pixels e normalizadas para valores entre 0 e 1. O conjunto foi dividido em 80% para treinamento (2.822 imagens) e 20% para validação (705 imagens).

**4.2 Arquitetura do Modelo**
Utilizou-se a rede neural convolucional VGG16, com pesos pré-treinados no banco de dados ImageNet, por meio da técnica de Transfer Learning. As camadas convolucionais foram mantidas com seus pesos originais, e apenas a camada final de classificação foi substituída.

**4.3 Treinamento**
O modelo foi compilado utilizando o otimizador Adam e a função de perda binary crossentropy. O treinamento foi realizado em 10 épocas.## 📌 5. RESULTADOS E DISCUSSÃO

```python
import os, glob, kagglehub
from tensorflow.keras import layers, models
from tensorflow.keras.applications import VGG16
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt

base = kagglehub.dataset_download("andrewmvd/leukemia-classification")
caminho = glob.glob(f"{base}/**/fold_0", recursive=True)[0]
print("Conferindo:", os.listdir(caminho))

train_gen = ImageDataGenerator(rescale=1./255, validation_split=0.2)
train = train_gen.flow_from_directory(caminho, target_size=(224,224), batch_size=32, class_mode='binary', subset='training')
val = train_gen.flow_from_directory(caminho, target_size=(224,224), batch_size=32, class_mode='binary', subset='validation')
print(f"DEU CERTO: {train.class_indices}")

base_model = VGG16(weights='imagenet', include_top=False, input_shape=(224,224,3))
base_model.trainable = False
model = models.Sequential([base_model, layers.GlobalAveragePooling2D(), layers.Dense(128, activation='relu'), layers.Dropout(0.5), layers.Dense(1, activation='sigmoid')])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

hist = model.fit(train, validation_data=val, epochs=10)

plt.plot(hist.history['accuracy'], label='Treino')
plt.plot(hist.history['val_accuracy'], label='Val')
plt.title('Acurácia - CORRETO'); plt.legend(); plt.show()
print(f"Final: {hist.history['val_accuracy'][-1]:.2%}")
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

val.reset()
y_pred_prob = model.predict(val, verbose=1)
y_pred = (y_pred_prob > 0.5).astype(int).flatten()
y_true = val.classes

cm = confusion_matrix(y_true, y_pred)
nomes_classes = ['ALL (Leucemia)', 'HEM (Normal)']

plt.figure(figsize=(6, 5), dpi=200)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=nomes_classes, yticklabels=nomes_classes)
plt.xlabel('Predito')
plt.ylabel('Real')
plt.title('Matriz de Confusão - VGG16')
plt.savefig('matriz_confusao_TCC.png', dpi=300, bbox_inches='tight')
plt.show()

print(classification_report(y_true, y_pred, target_names=nomes_classes))
```
